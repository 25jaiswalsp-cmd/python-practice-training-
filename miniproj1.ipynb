```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    'name':['A','B','C','D','E'],
    'Math':[23,45,67,77,67],
    'sci':[12,23,34,56,67],
    'eng':[12,34,46,8,88]
}
df = pd.DataFrame(data)
print(df)
```

```python

marks = df[['Math','sci','eng']].values
avg_marks=np.mean(marks,axis=1)
df['avg']=avg_marks
print(df)
```

```python
def grade(avg):
  if avg>=90:
    return 'A'
  elif avg>=75:
    return 'B'
  elif avg>=60:
    return 'C'
  elif avg>=40:
    return 'D'
  else:
    return 'E'

df['grade']=df['avg'].apply(grade)
df
```

```python
plt.figure()
plt.bar(df['name'],df['avg'])
plt.xlabel('student name')
plt.ylabel('avg marks')
plt.title('avg marks of students')
plt.show()
```

```python
df[['Math','sci','eng']].mean().plot(kind='bar')
plt.title('avg marks per subject')
plt.ylabel('marks')
plt.show()
```
