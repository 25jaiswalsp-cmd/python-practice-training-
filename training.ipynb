inheritance /hybrid inheritance

```python
#single level inheri
class A:
  def showA(self):
    print("showA func")
class B(A):
  def showB(self):
    print("showB func")


obj=B()
obj.showA()
obj.showB()
```

```python

class A:
  def features(self):
    print("showA func")
class B(A):
  def showB(self):
    print("showB func")
class C(A):
  def showC(self):
    print("showC func")


obj=B()
obj.features()
obj.showB()

obj1=C()
obj1.features()
obj1.showB()
obj1.showC()
```

```python

class A:
  def showA(self):
    print("showA func")
class B:
  def showB(self):
    print("showB func")
class C(A,B):
  def showC(self):
    print("showC func")


obj=C()
obj.showA()
obj.showB()
obj.showC()
```
