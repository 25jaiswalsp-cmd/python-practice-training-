```python
from abc import ABC, abstractmethod
class Test(ABC):
  @abstractmethod
  def m1(self):
    pass
  @abstractmethod
  def m2(self):
    pass
class mypython(Test):
    def m1(self):
      print("m1 method")
    @abstractmethod
    def m2(self):
      print("m2 method")
class child(mypython):
  def m2(self):
    print("child class")

m=child()
m.m1()
m.m2()
```

```python
class Test1:
  _b=None
  __c=None
  def __init__(self):
    self.a=101
    self._b=102
    self.__c=103
  def show1(self):
    print('a=',self.a)
    print('_b=',self._b)
    print('__c=',self.__c)
class Test2(Test1):
  def show2(self):
    print('a=',self.a)
    print('_b=',self._b)
class Test2(Test1):
  def show2(self):
    print('a=',self.a)
    print('_b=',self._b)
class Test3(Test2):
  def show3(self):
    print('a=',self.a)
    print('_b=',self._b)

t1=Test1()
t1.show1()
t2=Test2()
t2.show2()
t3=Test3()
t3.show3()
```

```python
import os,sys
fname=input("enter file name:")
if os.path.isfile(fname):
  print("file exist:", fname)
  f=open(fname,'r')
else:
  print("file is not:", fname)
  sys.exit(0)
lcount=wcount=ccount=0
for line in f:
  lcount=lcount+1
  ccount=ccount+len(line)
  words=line.split()
  wcount=wcount+len(words)
print("no.of lines",lcount)
print("no.of words",wcount)
print("no.of characters",ccount)
```
